```python

import torch
import transformers.pytorch_utils as pytorch_utils

def find_pruneable_heads_and_indices(
    heads,
    n_heads,
    head_size,
    already_pruned_heads
):
    mask = torch.ones(n_heads, head_size)

    heads = set(heads) - already_pruned_heads

    for head in heads:
        head -= sum(
            1 for h in already_pruned_heads if h < head
        )
        mask[head] = 0

    index = torch.arange(mask.numel())[
        mask.view(-1).bool()
    ]

    return heads, index.long()


pytorch_utils.find_pruneable_heads_and_indices = (
    find_pruneable_heads_and_indices
)

from src.models.encoder.dino import ViTModel


def compatible_get_head_mask(self, head_mask, num_hidden_layers):
    if head_mask is None:
        return [None] * num_hidden_layers

    return self._convert_head_mask_to_5d(
        head_mask,
        num_hidden_layers
    )


ViTModel.get_head_mask = compatible_get_head_mask

print("Transformers compatibility patch applied!")
```

```python

import importlib
import importlib.metadata as metadata
import torch

# Module name -> package distribution name
dependencies = {
    "transformers": "transformers",
    "diffusers": "diffusers",
    "accelerate": "accelerate",
    "safetensors": "safetensors",
    "huggingface_hub": "huggingface-hub",
    "timm": "timm",
    "einops": "einops",
    "omegaconf": "omegaconf",
    "open_clip": "open-clip-torch",

    "numpy": "numpy",
    "PIL": "Pillow",
    "cv2": "opencv-python-headless",
    "scipy": "scipy",
    "skimage": "scikit-image",
    "imageio": "imageio",
    "rembg": "rembg",
    "onnxruntime": "onnxruntime",

    "trimesh": "trimesh",
    "mcubes": "PyMCubes",
    "xatlas": "xatlas",
    "pymeshlab": "pymeshlab",
    "pygltflib": "pygltflib",
    "moderngl": "moderngl",
    "plyfile": "plyfile",

    "fastapi": "fastapi",
    "uvicorn": "uvicorn",
    "multipart": "python-multipart",
    "requests": "requests",
    "pydantic": "pydantic",
    "gradio": "gradio",

    "tqdm": "tqdm",
    "matplotlib": "matplotlib",
    "ninja": "ninja",

    "nvdiffrast.torch": "nvdiffrast",
}

results = {"PASS": [], "FAIL": []}

print("=" * 65)
print("SKETCH2SHAPE AI - DEPENDENCY AUDIT")
print("=" * 65)

for module_name, package_name in dependencies.items():
    try:
        module = importlib.import_module(module_name)

        try:
            version = metadata.version(package_name)
        except metadata.PackageNotFoundError:
            version = getattr(module, "__version__", "Version unavailable")

        results["PASS"].append(package_name)
        print(f"[OK]      {package_name:<25} {version}")

    except Exception as error:
        results["FAIL"].append((package_name, str(error)))
        print(f"[FAILED]  {package_name:<25} {str(error)[:100]}")

print("\n" + "=" * 65)
print("PYTORCH AND CUDA")
print("=" * 65)

print("PyTorch:", torch.__version__)
print("CUDA:", torch.version.cuda)
print("CUDA available:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
    print("GPU capability:", torch.cuda.get_device_capability(0))

print("\n" + "=" * 65)
print("AUDIT SUMMARY")
print("=" * 65)

print("Passed:", len(results["PASS"]))
print("Failed:", len(results["FAIL"]))

if results["FAIL"]:
    print("\nMissing or problematic dependencies:")
    for package, error in results["FAIL"]:
        print(f"- {package}: {error}")
else:
    print("\nAll checked dependencies imported successfully!")
```

```python
from pathlib import Path

search_roots = [
    PROJECT_DIR / "ml",
    PROJECT_DIR / "services",
    PROJECT_DIR / "backend",
    Path("/content/InstantMesh/src")
]

keywords = [
    "open_clip",
    "mcubes",
    "xatlas",
    "pymeshlab",
    "pygltflib",
    "moderngl",
    "plyfile"
]

for keyword in keywords:
    print(f"\n{'='*55}")
    print("PACKAGE:", keyword)
    print("="*55)

    found = False

    for root in search_roots:
        if not root.exists():
            continue

        for file in root.rglob("*.py"):
            try:
                lines = file.read_text(errors="ignore").splitlines()

                for line_no, line in enumerate(lines, 1):
                    if keyword.lower() in line.lower():
                        print(f"{file}:{line_no}: {line.strip()}")
                        found = True

            except Exception:
                pass

    if not found:
        print("No references found.")
```

```python
import time
import trimesh
from pathlib import Path

print("=" * 60)
print("SKETCH2SHAPE AI - END-TO-END VERIFICATION")
print("=" * 60)

# 1. Backend health
import requests

try:
    r = requests.get(
        "http://127.0.0.1:8000/api/health",
        timeout=10
    )
    print("\nBackend:", r.status_code)
    print("Health:", r.json())
except Exception as e:
    print("\nBackend FAILED:", e)

# 2. Latest backend logs
log_text = backend_log.read_text(errors="replace")

print("\n--- Generation log checks ---")

checks = {
    "ControlNet inference": "100%|",
    "3D generation completed": "3D generation completed",
    "Mesh validation": "Validating 3D mesh",
    "Pipeline completed": "Sketch2Shape pipeline completed",
    "API request successful": 'POST /api/generate HTTP/1.1" 200'
}

for name, keyword in checks.items():
    print(
        f"{name}:",
        "FOUND" if keyword in log_text else "NOT CONFIRMED"
    )

# 3. Find exported models
output_dirs = [
    Path(OUTPUT_DIR),
    Path(PROJECT_DIR) / "outputs",
    Path(PROJECT_DIR) / "examples" / "generated_outputs"
]

files = []

for folder in output_dirs:
    if folder.exists():
        files.extend(
            f for f in folder.rglob("*")
            if f.is_file() and f.suffix.lower() in [".glb", ".obj"]
        )

# Keep only recently generated files
files = sorted(
    set(files),
    key=lambda f: f.stat().st_mtime,
    reverse=True
)

print("\n--- Exported model verification ---")

for file in files[:5]:
    print(f"\nFile: {file.name}")
    print(f"Size: {file.stat().st_size / 1024:.2f} KB")

    try:
        mesh = trimesh.load(file, force="mesh")

        print("Vertices:", len(mesh.vertices))
        print("Faces:", len(mesh.faces))
        print("Non-empty geometry:", len(mesh.vertices) > 0)
        print("Watertight:", mesh.is_watertight)

    except Exception as e:
        print("FILE VALIDATION FAILED:", str(e))

print("\nVerification finished.")
```
